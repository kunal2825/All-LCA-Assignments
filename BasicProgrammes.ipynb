```python
### Simple Calculator

num1 = float(input("Enter first number"))
num2 = float(input("Enter second number"))

print("\nChoose an operation:, 1. Addition, 2. Subtraction, 3. Multiplication, 4. Division")

choice = input("Enter your choice (1-4):")

if choice == "1":
    print(num1 + num2)

elif choice == "2":
    print(num1 - num2)

elif choice == "3":
    print(num1 * num2)

elif choice == "4":
    print(num1 / num2)
```

```python
#Check if the given number is even or odd

x = int(input("Enter a number:"))

if x%2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
#Check which number is greatest among 3 numbers

x=int(input("Enter 1st number:"))
y=int(input("Enter 2nd number:"))
z=int(input("Enter 3rd number:"))
if x>y>z:
    print("x is greatest")
elif y>x>z:
    print("y is greatest")
else:
    print("z is greatest")
```

```python
#Program to print sum of all digits of a number

num = int(input("Enter a number:"))
sum = 0
while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```

```python
#Find factorial of given number

num = int(input("Enter a number:"))
f = 1
i = 1

while i <= num:
    f = f * i
    i = i + 1
    print("Factorial =", f)
```

```python
#Check if given number is prime or not

num = int(input("Enter a number:"))

for i in range(2,num):
    if num%i == 0:
        print("Not Prime")
        break
else:
    print("Prime")
```

```python
#Check whether a student has passed or failed and their grade

marks = []

for i in range(5):
    marks.append(int(input("Enter marks:")))

avg = sum(marks) / 5

if min(marks) < 30:
    print("Fail")
elif avg >= 90:
    print("Grade A+")
elif avg >= 80:
    print("Grade A")
elif avg >= 70:
    print("Grade B+")
elif avg >= 60:
    print("Grade B")
elif avg >= 50:
    print("Grade C+")
elif avg >= 40:
    print("Grade C")
else:
    print("Grade D")
```
