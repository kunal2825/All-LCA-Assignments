```python
#Create a list of 5 integers and display the list items

numbers = [10, 20, 30, 40, 50]

print("List items:", numbers)

print("First number:", numbers[0])
print("Second number:", numbers[1])
print("Third number:", numbers[2])
print("Fourth number:", numbers[3])
print("Fifth number:", numbers[4])
```

```python
#Append new item to the end of list

numbers = [10, 20, 30, 40, 50]
print("Original List", numbers)
numbers.append(60)
print("After Appending:", numbers)
```

```python
#Reverse the order of items in the list

numbers = [10, 20, 30, 40, 50]
print("Original List", numbers)
numbers.reverse()
print("Reversed List", numbers)
```

```python
#Print the number of occurences of a specified element in a list

numbers = [10, 20, 10, 30, 10, 40, 50]
element = int(input("Enter the element to scratch:"))
count = numbers.count(element)
print("Number of occurences:", count)
```

```python
#Append the items of list 1 to list 2 in the front

list1=[10, 20, 30, 40]
list2=[50, 60, 70, 80]

list2 = list1 + list2

print("List1:", list1)
print("List2:", list2)
```

```python
#Insert a new item before the second element in an existing list

numbers = [10, 20, 30, 40, 50]
print("Original List:", numbers)
numbers.insert(1, 15)
print("After Insertion:",numbers)
```

```python
#Remove the item from a specified index in a list

numbers = [10, 20, 30, 40, 50]
index = int(input("Enter the index to remove:"))
numbers.pop(index)
print("List after removal:",numbers)
```

```python
#Remove the first occurence of a specified element from a list

numbers = [10, 20, 30, 40, 50]
element = int(input("Enter the element to remove:"))
numbers.remove(element)
print("List after removal:",numbers)
```
